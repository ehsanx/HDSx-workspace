```python

import pandas as pd

nhanes_py = pd.read_csv("../../../examples/nhanes-equity/data/nhanes_equity_v6.csv", low_memory=False)
nhanes_py.head()

python_summary = (
    nhanes_py.dropna(subset=["BMI", "IncomeGroup", "Gender"])
    .groupby(["IncomeGroup", "Gender"], as_index=False)
    .agg(
        n=("BMI", "size"),
        mean_bmi=("BMI", "mean"),
        sd_bmi=("BMI", "std"),
    )
    .assign(
        mean_bmi=lambda summary: summary["mean_bmi"].round(2),
        sd_bmi=lambda summary: summary["sd_bmi"].round(2),
    )
    .sort_values(["IncomeGroup", "Gender"])
    .reset_index(drop=True)
)

```
